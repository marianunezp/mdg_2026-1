```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
from sqlalchemy import text
```

```python
DB_HOST = "localhost"
DB_PORT = 5432
DB_NAME = "censo_rm_2017"
DB_USER = "postgres"
DB_PASSWORD = "postgres"

engine = create_engine(
    f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

print("Conexión exitosa al servidor PostgreSQL")
```

## 4. Resultados

### 4.1 Cálculo de los indicadores a nivel comunal

Se seleccionaron tres indicadores relacionados con la estructura etaria de la población: porcentaje de población infantil de 0 a 14 años, porcentaje de adultos mayores de 65 años o más e índice de envejecimiento.

El porcentaje de población infantil y el porcentaje de adultos mayores indican la participación de cada grupo dentro de la población total de cada comuna. Por su parte, el índice de envejecimiento expresa la cantidad de adultos mayores de 65 años o más por cada 100 personas de 0 a 14 años. Un valor elevado representa una población relativamente más envejecida, mientras que un valor bajo indica una mayor presencia de población infantil.

```python
sql_indicadores = """
SELECT
    c.nom_comuna,

    ROUND(
        100.0 * COUNT(*) FILTER (WHERE p.p09 BETWEEN 0 AND 14)
        / NULLIF(COUNT(*), 0),
        2
    ) AS ptje_poblacion_infantil,

    ROUND(
        100.0 * COUNT(*) FILTER (WHERE p.p09 >= 65)
        / NULLIF(COUNT(*), 0),
        2
    ) AS ptje_adultos_mayores,

    ROUND(
        100.0 * COUNT(*) FILTER (WHERE p.p09 >= 65)
        / NULLIF(
            COUNT(*) FILTER (WHERE p.p09 BETWEEN 0 AND 14),
            0
        ),
        2
    ) AS indice_envejecimiento

FROM personas AS p

JOIN hogares AS h
    ON p.hogar_ref_id = h.hogar_ref_id

JOIN viviendas AS v
    ON h.vivienda_ref_id = v.vivienda_ref_id

JOIN zonas AS z
    ON v.zonaloc_ref_id = z.zonaloc_ref_id

JOIN comunas AS c
    ON z.codigo_comuna = c.codigo_comuna

JOIN provincias AS pr
    ON c.provincia_ref_id = pr.provincia_ref_id

WHERE pr.nom_provincia = 'SANTIAGO'
   OR c.nom_comuna IN ('PUENTE ALTO', 'SAN BERNARDO')

GROUP BY c.nom_comuna
ORDER BY c.nom_comuna;
"""

df_indicadores = pd.read_sql(sql_indicadores, engine)

df_indicadores
```

### 4.2 Análisis general de los indicadores comunales

Los resultados muestran diferencias en la estructura etaria de las comunas del Gran Santiago. En la mayoría de las comunas, el porcentaje de población infantil supera al porcentaje de adultos mayores. Sin embargo, algunas comunas del sector oriente y centro presentan una población relativamente más envejecida.

El índice de envejecimiento permite observar esta diferencia con mayor claridad, ya que relaciona directamente la cantidad de adultos mayores con la población infantil. De esta manera, es posible reconocer comunas con una estructura más joven y otras en las que existe una mayor presencia relativa de adultos mayores.

### 4.3 Gráfico de población infantil

El siguiente gráfico presenta el porcentaje de población infantil de 0 a 14 años en las comunas del Gran Santiago, ordenado de mayor a menor.

```python
df_infantil = df_indicadores.sort_values(
    "ptje_poblacion_infantil",
    ascending=True
)

plt.figure(figsize=(10, 12))

plt.barh(
    df_infantil["nom_comuna"],
    df_infantil["ptje_poblacion_infantil"],
    color="cornflowerblue"
)

plt.title(
    "Porcentaje de población infantil por comuna\nGran Santiago, Censo 2017"
)
plt.xlabel("Población infantil de 0 a 14 años (%)")
plt.ylabel("Comuna")
plt.grid(axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

#### Análisis del gráfico de población infantil

El gráfico muestra diferencias territoriales en la presencia de población infantil en el Gran Santiago. La Pintana presenta el porcentaje más alto, con aproximadamente un 23 %, seguida por Huechuraba, Lo Barnechea, Quilicura y San Bernardo. ### 4.7 Mapas comunales de los indicadores

Para representar espacialmente los tres indicadores, los resultados se relacionaron con la capa geográfica `dpa.comunas_rm_shp`. Esta unión permite visualizar la distribución territorial de la población infantil, los adultos mayores y el índice de envejecimiento en las comunas del Gran Santiago.

```python
sql_mapas = """
WITH agg AS (
    SELECT
        c.nom_comuna,

        ROUND(
            100.0 * COUNT(*) FILTER (WHERE p.p09 BETWEEN 0 AND 14)
            / NULLIF(COUNT(*), 0),
            2
        ) AS ptje_poblacion_infantil,

        ROUND(
            100.0 * COUNT(*) FILTER (WHERE p.p09 >= 65)
            / NULLIF(COUNT(*), 0),
            2
        ) AS ptje_adultos_mayores,

        ROUND(
            100.0 * COUNT(*) FILTER (WHERE p.p09 >= 65)
            / NULLIF(
                COUNT(*) FILTER (WHERE p.p09 BETWEEN 0 AND 14),
                0
            ),
            2
        ) AS indice_envejecimiento

    FROM public.personas AS p
    JOIN public.hogares AS h
        ON p.hogar_ref_id = h.hogar_ref_id
    JOIN public.viviendas AS v
        ON h.vivienda_ref_id = v.vivienda_ref_id
    JOIN public.zonas AS z
        ON v.zonaloc_ref_id = z.zonaloc_ref_id
    JOIN public.comunas AS c
        ON z.codigo_comuna = c.codigo_comuna
    JOIN public.provincias AS pr
        ON c.provincia_ref_id = pr.provincia_ref_id

    WHERE pr.nom_provincia = 'SANTIAGO'
       OR c.nom_comuna IN ('PUENTE ALTO', 'SAN BERNARDO')

    GROUP BY c.nom_comuna
)

SELECT
    a.nom_comuna,
    shp.geom,
    a.ptje_poblacion_infantil,
    a.ptje_adultos_mayores,
    a.indice_envejecimiento

FROM agg AS a
JOIN dpa.comunas_rm_shp AS shp
    ON UPPER(shp.nom_comuna) = a.nom_comuna;
"""

gdf_mapas = gpd.read_postgis(
    sql_mapas,
    engine,
    geom_col="geom"
)

print("Cantidad de comunas:", len(gdf_mapas))

gdf_mapas[
    [
        "nom_comuna",
        "ptje_poblacion_infantil",
        "ptje_adultos_mayores",
        "indice_envejecimiento"
    ]
].head()
```

#### 4.7.1 Mapa de población infantil

El mapa muestra que la población infantil no se distribuye de manera uniforme en el Gran Santiago. Los tonos más oscuros se concentran principalmente en comunas periféricas, como La Pintana, Puente Alto, San Bernardo, Quilicura y Lo Barnechea, indicando una mayor proporción de habitantes entre 0 y 14 años.

En cambio, las tonalidades más claras se observan en comunas centrales y del sector oriente, como Santiago, Providencia y Ñuñoa, donde la población infantil representa un porcentaje menor. Este patrón evidencia diferencias territoriales en la estructura etaria y en las necesidades de servicios educativos, recreativos y de salud infantil.

## 5. Indicadores demográficos por zona censal

En esta sección se calculan el porcentaje de población infantil, el porcentaje de adultos mayores y el índice de envejecimiento a nivel de zona censal. Esta escala permite observar diferencias internas que no se distinguen al utilizar solamente los resultados comunales.

```python
sql_indicadores_zona = """
SELECT
    c.nom_comuna,
  z.geocodigo AS id_zona,
    ROUND(
        100.0 * COUNT(*) FILTER (WHERE p.p09 BETWEEN 0 AND 14)
        / NULLIF(COUNT(*), 0),
        2
    ) AS ptje_poblacion_infantil,

    ROUND(
        100.0 * COUNT(*) FILTER (WHERE p.p09 >= 65)
        / NULLIF(COUNT(*), 0),
        2
    ) AS ptje_adultos_mayores,

    ROUND(
        100.0 * COUNT(*) FILTER (WHERE p.p09 >= 65)
        / NULLIF(
            COUNT(*) FILTER (WHERE p.p09 BETWEEN 0 AND 14),
            0
        ),
        2
    ) AS indice_envejecimiento

FROM public.personas AS p

JOIN public.hogares AS h
    ON p.hogar_ref_id = h.hogar_ref_id

JOIN public.viviendas AS v
    ON h.vivienda_ref_id = v.vivienda_ref_id

JOIN public.zonas AS z
    ON v.zonaloc_ref_id = z.zonaloc_ref_id

JOIN public.comunas AS c
    ON z.codigo_comuna = c.codigo_comuna

JOIN public.provincias AS pr
    ON c.provincia_ref_id = pr.provincia_ref_id

WHERE pr.nom_provincia = 'SANTIAGO'
   OR c.nom_comuna IN ('PUENTE ALTO', 'SAN BERNARDO')

GROUP BY
    c.nom_comuna,
    z.geocodigo

ORDER BY
    c.nom_comuna,
    z.geocodigo;
"""

df_indicadores_zona = pd.read_sql(
    sql_indicadores_zona,
    engine
)

print("Cantidad de zonas censales:", len(df_indicadores_zona))

df_indicadores_zona.head(10)
```

```python
print("Valores nulos por columna:")
print(df_indicadores_zona.isnull().sum())

print("\nResumen estadístico:")
display(
    df_indicadores_zona[
        [
            "ptje_poblacion_infantil",
            "ptje_adultos_mayores",
            "indice_envejecimiento"
        ]
    ].describe().round(2)
)
```

```python
sql_geometria_zonas = """
SELECT
    geocodigo::bigint::text AS id_zona,
    nom_comuna,
    nom_zona,
    geom
FROM dpa.zonas_censales_rm
WHERE UPPER(nom_provin) = 'SANTIAGO'
   OR UPPER(nom_comuna) IN ('PUENTE ALTO', 'SAN BERNARDO');
"""

with engine.connect() as conexion:
    gdf_zonas = gpd.read_postgis(
        text(sql_geometria_zonas),
        conexion,
        geom_col="geom"
    )

# Limpiar y convertir ambos identificadores al mismo formato
df_indicadores_zona["id_zona"] = (
    pd.to_numeric(df_indicadores_zona["id_zona"], errors="coerce")
    .astype("Int64")
    .astype(str)
)

gdf_zonas["id_zona"] = (
    pd.to_numeric(gdf_zonas["id_zona"], errors="coerce")
    .astype("Int64")
    .astype(str)
)


# Unir las geometrías con los tres indicadores
gdf_indicadores_zona = gdf_zonas.merge(
    df_indicadores_zona,
    on="id_zona",
    how="inner",
    suffixes=("_mapa", "_datos")
)

print("Zonas geográficas obtenidas:", len(gdf_zonas))
print("Zonas con indicadores:", len(df_indicadores_zona))
print("Zonas unidas con indicadores:", len(gdf_indicadores_zona))

gdf_indicadores_zona[
    [
        "id_zona",
        "nom_comuna_datos",
        "ptje_poblacion_infantil",
        "ptje_adultos_mayores",
        "indice_envejecimiento"
    ]
].head(10)
```

```python
# Zonas geográficas que no encontraron indicadores
zonas_sin_indicadores = gdf_zonas[
    ~gdf_zonas["id_zona"].isin(df_indicadores_zona["id_zona"])
]

print("Zonas geográficas sin indicadores:", len(zonas_sin_indicadores))

display(
    zonas_sin_indicadores[
        ["id_zona", "nom_comuna", "nom_zona"]
    ]
)

# Indicadores que no encontraron geometría
indicadores_sin_geometria = df_indicadores_zona[
    ~df_indicadores_zona["id_zona"].isin(gdf_zonas["id_zona"])
]

print(
    "Registros de indicadores sin geometría:",
    len(indicadores_sin_geometria)
)

display(
    indicadores_sin_geometria[
        ["id_zona", "nom_comuna"]
    ].head(40)
)
```

```python
# Identificar zonas censales con valores extremos
columnas_resultado = [
    "id_zona",
    "nom_comuna_datos",
    "ptje_poblacion_infantil",
    "ptje_adultos_mayores",
    "indice_envejecimiento"
]

print("5 zonas con mayor porcentaje de población infantil:")
display(
    gdf_indicadores_zona.nlargest(
        5, "ptje_poblacion_infantil"
    )[columnas_resultado]
)

print("5 zonas con menor porcentaje de población infantil:")
display(
    gdf_indicadores_zona.nsmallest(
        5, "ptje_poblacion_infantil"
    )[columnas_resultado]
)

print("5 zonas con mayor porcentaje de adultos mayores:")
display(
    gdf_indicadores_zona.nlargest(
        5, "ptje_adultos_mayores"
    )[columnas_resultado]
)

print("5 zonas con mayor índice de envejecimiento:")
display(
    gdf_indicadores_zona.nlargest(
        5, "indice_envejecimiento"
    )[columnas_resultado]
)
```

### 5.1 Mapa de población infantil por zona censal

El siguiente mapa representa la distribución del porcentaje de población de 0 a 14 años en las zonas censales del Gran Santiago. Los tonos azules más oscuros indican una mayor proporción de población infantil.

#### Análisis del mapa de población infantil por zona censal

La población infantil presenta diferencias importantes entre las zonas censales del Gran Santiago. Los porcentajes más altos se registran en las zonas 13201051004 de Puente Alto (35,61 %), 13115011011 de Lo Barnechea (35,39 %) y 13113071005 de La Reina (34,03 %). En contraste, algunas zonas de Pudahuel y La Pintana registran valores cercanos o iguales a 0 %, posiblemente debido a una población infantil inexistente o muy reducida en esos sectores. Estos resultados demuestran que dentro de una misma comuna pueden existir realidades demográficas diferentes, las cuales no se distinguen al observar únicamente el promedio comunal.

### 5.3 Mapa del índice de envejecimiento por zona censal

El siguiente mapa representa la cantidad de adultos mayores de 65 años o más por cada 100 personas de 0 a 14 años en las zonas censales del Gran Santiago. Los tonos morados más oscuros indican un mayor índice de envejecimiento.

```python
fig, ax = plt.subplots(figsize=(12, 12))

gdf_indicadores_zona.plot(
    column="indice_envejecimiento",
    cmap="Purples",
    linewidth=0.1,
    edgecolor="gray",
    legend=True,
    ax=ax,
    missing_kwds={
        "color": "lightgray",
        "label": "Sin información"
    }
)

ax.set_title(
    "Índice de envejecimiento por zona censal\nGran Santiago",
    fontsize=15
)

ax.set_axis_off()

plt.tight_layout()
plt.show()
```

#### Análisis del mapa del índice de envejecimiento por zona censal

El índice de envejecimiento alcanza su valor más alto en la zona 13123011001 de Providencia, con 386,21 adultos mayores por cada 100 niños. También destacan la zona 13132011005 de Vitacura (267,12), la zona 13123031002 de Providencia (253,49) y zonas de Las Condes con valores cercanos a 240. Estos resultados reflejan sectores donde la población adulta mayor supera ampliamente a la población infantil, lo que podría generar una mayor necesidad de servicios de salud, accesibilidad, cuidados y espacios adaptados.

## 6. Conclusiones

En respuesta a la pregunta de investigación, los resultados muestran que la población infantil y los adultos mayores no se distribuyen uniformemente en el Gran Santiago. A nivel comunal, la población infantil presenta una mayor presencia relativa en comunas periféricas como La Pintana, Quilicura, Puente Alto y San Bernardo, mientras que los mayores niveles de envejecimiento se concentran principalmente en comunas centrales y del sector oriente, como Providencia, Vitacura y Las Condes.

El análisis por zona censal permitió observar diferencias internas que los promedios comunales pueden ocultar. Por ejemplo, zonas específicas de Puente Alto, Lo Barnechea y La Reina presentan altos porcentajes de población infantil, mientras que algunas zonas de San Bernardo y Vitacura registran elevados porcentajes de adultos mayores. Los mayores índices de envejecimiento se observan principalmente en zonas de Providencia, Vitacura y Las Condes.

Estas diferencias pueden relacionarse con la expansión urbana, la antigüedad de los barrios, la permanencia de población de mayor edad y la presencia de familias jóvenes en sectores de crecimiento residencial. Por lo tanto, la escala censal entrega información más detallada para orientar territorialmente servicios educativos, espacios recreativos, atención de salud, accesibilidad y cuidados para adultos mayores.

En conclusión, el uso conjunto de SQL, Python, gráficos y mapas temáticos permitió responder la pregunta planteada e identificar necesidades demográficas diferentes dentro del Gran Santiago. Esto demuestra la importancia de considerar tanto la escala comunal como la censal para apoyar una planificación territorial más precisa.
